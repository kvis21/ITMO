```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Определяем функции
def f_original(x):
    return (18 * x) / (x**4 + 10)

def f_quadratic(x):
    return 0.372 + 1.172 * x - 0.325 * x**2

def f_linear(x):
    return 0.432 + 0.032 * x

# 2. Генерируем точки по оси X (500 точек для идеальной плавности)
x = np.linspace(-1, 5, 500)

# 3. Считаем значения Y для каждого графика
y_original = f_original(x)
y_quadratic = f_quadratic(x)
y_linear = f_linear(x)

# 4. Настраиваем размер и сетку графика
plt.figure(figsize=(10, 6))
plt.grid(True, linestyle="--", alpha=0.5)

# 5. Строим линии (используем LaTeX-разметку для красивых формул в легенде)
plt.plot(x, y_original, label=r'Исходная: $y = \frac{18x}{x^4+10}$', color='blue', linewidth=2.5)
plt.plot(x, y_quadratic, label=r'Квадратичное: $y = 0.372 + 1.172x - 0.325x^2$', color='red', linestyle='--')
plt.plot(x, y_linear, label=r'Линейное: $y = 0.432 + 0.032x$', color='green', linestyle='-.')

# 6. Оформляем подписи и заголовок
plt.title('Сравнение исходной функции и её приближений', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Ось X', fontsize=12)
plt.ylabel('Ось Y', fontsize=12)

# Добавляем легенду (автоматически выберет лучшее место на графике)
plt.legend(fontsize=11, loc='best')

# 7. Отображаем результат
plt.show()
```
